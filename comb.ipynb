```python
from pathlib import Path

# Determine project root (works in scripts and in Jupyter)
try:
    PROJECT_ROOT = Path(__file__).resolve().parent.parent
except NameError:
    PROJECT_ROOT = Path.cwd()  # Jupyter: cwd is typically the project root

SRC_DIR = PROJECT_ROOT / "src" / "pm_agent"
OUT_FILE = PROJECT_ROOT / "combined_pm_agent.py"

def main():
    src = SRC_DIR
    out = OUT_FILE

    print(f"PROJECT_ROOT={PROJECT_ROOT}")
    print(f"SRC_DIR={src}")
    print(f"OUT_FILE={out}")

    if not src.exists() or not src.is_dir():
        print(f"Source directory not found: {src}")
        return

    py_files = sorted([p for p in src.rglob("*.py") if p.is_file() and p.suffix == ".py"])
    # avoid writing the output into itself if placed under the source tree
    py_files = [p for p in py_files if p.resolve() != out.resolve()]

    if not py_files:
        print("No .py files found to combine.")
        return

    header = (
        "# Combined Python file\n"
        f"# Source directory: {src}\n"
        "# Files included (in order):\n"
    )
    header += "\n".join(f"# - {p.relative_to(src)}" for p in py_files)
    header += "\n\n"

    out.parent.mkdir(parents=True, exist_ok=True)
    with out.open("w", encoding="utf-8") as f:
        f.write(header)
        for p in py_files:
            f.write(f"# ---------- FILE: {p.relative_to(src)} ----------\n")
            try:
                content = p.read_text(encoding="utf-8")
            except Exception as e:
                content = f"# Error reading file {p}: {e}\n"
            f.write(content.rstrip() + "\n\n")

    print(f"Wrote combined file to: {out}")

if __name__ == "__main__":
    main()
```
